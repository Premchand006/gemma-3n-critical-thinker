```python
# Add your Hugging Face token as a Colab secret named HF_TOKEN (key icon in the left bar).
import os
from google.colab import userdata
os.environ["HF_TOKEN"] = userdata.get("HF_TOKEN")
```

```python
from cogtools.tools import CognitiveToolsAgent, HFBackend

backend = HFBackend(
    "unsloth/gemma-3n-E4B-it",
    adapter="anhsirk/gemma-3n-critical-thinker",  # None for the base model
    load_in_4bit=True,
)
agent = CognitiveToolsAgent(backend, max_turns=8)
```

```python
trace = agent.solve("A rectangle has a perimeter of 30 cm and a length of 9 cm. What is its area in square centimetres?")
print(trace.to_markdown())
```
